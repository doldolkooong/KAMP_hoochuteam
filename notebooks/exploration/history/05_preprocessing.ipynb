```python
# 원본 변수는 그대로 유지
X_cols_cn7 = train_cn7.columns.drop('PassOrFail')

# 하나라도 |z| > 3이면 1
train_cn7['is_outlier'] = (
    train_cn7[X_cols_cn7].abs() > 3
).any(axis=1).astype(int)

# 한 행에서 이상치인 공정변수가 몇 개인지
train_cn7['outlier_count'] = (
    train_cn7[X_cols_cn7].abs() > 3
).sum(axis=1)
```

```python
for col in X_cols_cn7:
    train_cn7[f'{col}_outlier'] = (
        train_cn7[col].abs() > 3
    ).astype(int)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    average_precision_score,
    confusion_matrix,
    classification_report
)

# X / y 분리
X_cn7 = train_cn7.drop(columns='PassOrFail')
y_cn7 = train_cn7['PassOrFail']

# Stratified Split
X_train_cn7, X_test_cn7, y_train_cn7, y_test_cn7 = train_test_split(
    X_cn7,
    y_cn7,
    test_size=0.2,
    stratify=y_cn7,
    random_state=42
)

# Logistic Regression
logistic_cn7 = LogisticRegression(
    class_weight='balanced',
    max_iter=1000,
    random_state=42
)

logistic_cn7.fit(X_train_cn7, y_train_cn7)

# 예측
pred_cn7 = logistic_cn7.predict(X_test_cn7)
proba_cn7 = logistic_cn7.predict_proba(X_test_cn7)[:, 1]
```

```python
print("[CN7 Logistic Regression]")

print("Accuracy :", accuracy_score(y_test_cn7, pred_cn7))
print("Precision:", precision_score(y_test_cn7, pred_cn7, zero_division=0))
print("Recall   :", recall_score(y_test_cn7, pred_cn7, zero_division=0))
print("F1       :", f1_score(y_test_cn7, pred_cn7, zero_division=0))
print("PR-AUC   :", average_precision_score(y_test_cn7, proba_cn7))

print("\nConfusion Matrix")
print(confusion_matrix(y_test_cn7, pred_cn7))

print("\nClassification Report")
print(classification_report(y_test_cn7, pred_cn7, digits=4))
```

```python
from xgboost import XGBClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    average_precision_score,
    confusion_matrix,
    classification_report
)

# 클래스 불균형 비율
scale_pos_weight_cn7 = (
    (y_train_cn7 == 0).sum() /
    (y_train_cn7 == 1).sum()
)

print("scale_pos_weight:", scale_pos_weight_cn7)

xgb_cn7 = XGBClassifier(
    n_estimators=300,
    max_depth=3,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    scale_pos_weight=scale_pos_weight_cn7,
    eval_metric='logloss',
    random_state=42,
    n_jobs=-1
)

xgb_cn7.fit(X_train_cn7, y_train_cn7)

pred_xgb_cn7 = xgb_cn7.predict(X_test_cn7)
proba_xgb_cn7 = xgb_cn7.predict_proba(X_test_cn7)[:, 1]
```
