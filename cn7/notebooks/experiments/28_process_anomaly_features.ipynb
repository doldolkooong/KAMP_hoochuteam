```python
import sys

sys.path.append("..")

from common.metrics import metrics
from common.preprocessing import preprocessing_cn7
from common.train_test_split import split_data
from common.model_composition import model_composition
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 데이터 불러오기
train_cn7 = pd.read_csv(r"C:\Users\Administrator\Desktop\kamp\1. 사출성형기 AI 데이터셋\1. 사출성형기 AI 데이터셋\moldset_labeled_cn7.csv")
```

```python
X_cn7, y_cn7 = preprocessing_cn7(train_cn7)

print("X shape:", X_cn7.shape)
print("y shape:", y_cn7.shape)
```

### Mean_Abs_Deviation

```python
X_cn7_fe12 = X_cn7.copy()

X_cn7_fe12["Mean_Abs_Deviation"] = (
    X_cn7_fe12.abs().mean(axis=1)
)

X_train_cn7_fe12, X_test_cn7_fe12, y_train_cn7_fe12, y_test_cn7_fe12 = split_data(
    X_cn7_fe12,
    y_cn7
)

model_composition(
    X_train_cn7_fe12,
    X_test_cn7_fe12,
    y_train_cn7_fe12,
    y_test_cn7_fe12
)
```

### Max_Abs_Deviation


```python
X_cn7_fe13 = X_cn7.copy()

X_cn7_fe13["Max_Abs_Deviation"] = (
    X_cn7_fe13.abs().max(axis=1)
)

X_train_cn7_fe13, X_test_cn7_fe13, y_train_cn7_fe13, y_test_cn7_fe13 = split_data(
    X_cn7_fe13,
    y_cn7
)

model_composition(
    X_train_cn7_fe13,
    X_test_cn7_fe13,
    y_train_cn7_fe13,
    y_test_cn7_fe13
)
```

### Deviation_Count_2Sigma

```python
X_cn7_fe14 = X_cn7.copy()

X_cn7_fe14["Deviation_Count_2Sigma"] = (
    (X_cn7_fe14.abs() > 2).sum(axis=1)
)

X_train_cn7_fe14, X_test_cn7_fe14, y_train_cn7_fe14, y_test_cn7_fe14 = split_data(
    X_cn7_fe14,
    y_cn7
)

model_composition(
    X_train_cn7_fe14,
    X_test_cn7_fe14,
    y_train_cn7_fe14,
    y_test_cn7_fe14
)
```

### Deviation_Count_3Sigma

```python
X_cn7_fe15 = X_cn7.copy()

X_cn7_fe15["Deviation_Count_3Sigma"] = (
    (X_cn7_fe15.abs() > 3).sum(axis=1)
)

X_train_cn7_fe15, X_test_cn7_fe15, y_train_cn7_fe15, y_test_cn7_fe15 = split_data(
    X_cn7_fe15,
    y_cn7
)

model_composition(
    X_train_cn7_fe15,
    X_test_cn7_fe15,
    y_train_cn7_fe15,
    y_test_cn7_fe15
)
```
