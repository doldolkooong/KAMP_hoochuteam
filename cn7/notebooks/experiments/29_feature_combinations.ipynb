```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 데이터 불러오기
train_cn7 = pd.read_csv(r"C:\Users\Administrator\Desktop\kamp\1. 사출성형기 AI 데이터셋\1. 사출성형기 AI 데이터셋\moldset_labeled_cn7.csv")
```

```python
X_cn7, y_cn7 = preprocessing_cn7(train_cn7)

print("X shape:", X_cn7.shape)
print("y shape:", y_cn7.shape)
```

조합 1
- Process_Time_Sum → 시간 공정 전체 패턴
- Screw_RPM_Mean → 스크류 회전속도 수준
- Barrel_Temp_Std → 배럴 온도 분포의 불균일성
- Max_Abs_Deviation → 전체 공정 중 최대 이탈 정도

```python
X_cn7_combo1 = X_cn7.copy()

# 1. 시간
X_cn7_combo1["Process_Time_Sum"] = (
    X_cn7_combo1["Injection_Time"]
    + X_cn7_combo1["Filling_Time"]
    + X_cn7_combo1["Plasticizing_Time"]
    + X_cn7_combo1["Cycle_Time"]
    + X_cn7_combo1["Clamp_Close_Time"]
)

# 2. RPM
X_cn7_combo1["Screw_RPM_Mean"] = (
    X_cn7_combo1["Max_Screw_RPM"]
    + X_cn7_combo1["Average_Screw_RPM"]
) / 2

# 3. 배럴 온도
barrel_cols = [
    "Barrel_Temperature_1",
    "Barrel_Temperature_2",
    "Barrel_Temperature_3",
    "Barrel_Temperature_4",
    "Barrel_Temperature_5",
    "Barrel_Temperature_6"
]

X_cn7_combo1["Barrel_Temp_Std"] = (
    X_cn7_combo1[barrel_cols].std(axis=1)
)

# 4. 전체 공정 최대 이탈
# 파생변수들이 추가되기 전 원래 X_cn7 기준으로 계산
X_cn7_combo1["Max_Abs_Deviation"] = (
    X_cn7.abs().max(axis=1)
)

# Train / Test Split
X_train_cn7_combo1, X_test_cn7_combo1, y_train_cn7_combo1, y_test_cn7_combo1 = split_data(
    X_cn7_combo1,
    y_cn7
)

# 5개 모델 비교
model_composition(
    X_train_cn7_combo1,
    X_test_cn7_combo1,
    y_train_cn7_combo1,
    y_test_cn7_combo1
)
```
