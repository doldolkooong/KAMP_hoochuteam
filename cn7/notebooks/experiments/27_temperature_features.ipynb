### Barrel_Temp_Mean

```python
X_cn7_fe10 = X_cn7.copy()

barrel_cols = [
    "Barrel_Temperature_1",
    "Barrel_Temperature_2",
    "Barrel_Temperature_3",
    "Barrel_Temperature_4",
    "Barrel_Temperature_5",
    "Barrel_Temperature_6"
]

X_cn7_fe10["Barrel_Temp_Mean"] = (
    X_cn7_fe10[barrel_cols].mean(axis=1)
)

X_train_cn7_fe10, X_test_cn7_fe10, y_train_cn7_fe10, y_test_cn7_fe10 = split_data(
    X_cn7_fe10,
    y_cn7
)

model_composition(
    X_train_cn7_fe10,
    X_test_cn7_fe10,
    y_train_cn7_fe10,
    y_test_cn7_fe10
)
```

```python
X_cn7_fe11 = X_cn7.copy()

barrel_cols = [
    "Barrel_Temperature_1",
    "Barrel_Temperature_2",
    "Barrel_Temperature_3",
    "Barrel_Temperature_4",
    "Barrel_Temperature_5",
    "Barrel_Temperature_6"
]

X_cn7_fe11["Barrel_Temp_Std"] = (
    X_cn7_fe11[barrel_cols].std(axis=1)
)

X_train_cn7_fe11, X_test_cn7_fe11, y_train_cn7_fe11, y_test_cn7_fe11 = split_data(
    X_cn7_fe11,
    y_cn7
)

model_composition(
    X_train_cn7_fe11,
    X_test_cn7_fe11,
    y_train_cn7_fe11,
    y_test_cn7_fe11
)
```
